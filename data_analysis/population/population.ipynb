```python
import zipfile
import pandas as pd
```

```python
zip_path = 'settlements.zip'
```

```python
with zipfile.ZipFile(zip_path, 'r') as zip_file:
    # print(zip_file.namelist())
    data_file = zip_file.namelist()[1]

    with zip_file.open(data_file) as f:
        df = pd.read_csv(f)
```

```python
print(df.head())
```

```python
df['population'] = pd.to_numeric(df['population'], errors='coerce')
df = df.dropna(subset=['population'])
df = df[df['population'] > 0]
```

```python
median = df['population'].median()
mean = df['population'].mean()
```

```python
print(f"Median population: {median:,.0f}")
print(f'Mean population: {mean:,.0f}')
```

```python
max_population = df['population'].max()
print(f'Max population: {max_population:,.0f}')
```

```python
largest_place_row = df.loc[df['population'].idxmax()]
place_name = largest_place_row['settlement']
print(place_name)
```
